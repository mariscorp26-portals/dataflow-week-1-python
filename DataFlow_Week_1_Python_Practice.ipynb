## Assessment 5

```python
# Assessment 5 - FizzBuzz

# Print numbers from 1 to 50.
# Multiples of 3 = Fizz
# Multiples of 5 = Buzz
# Multiples of both = FizzBuzz

for number in range(1, 51):

    if number % 3 == 0 and number % 5 == 0:
        print("FizzBuzz")

    elif number % 3 == 0:
        print("Fizz")

    elif number % 5 == 0:
        print("Buzz")

    else:
        print(number)
```

## Assessment 7

```python
# Assessment 7 - Return only even numbers

def get_even_numbers(numbers):
    even_numbers = []

    for number in numbers:
        if number % 2 == 0:
            even_numbers.append(number)

    return even_numbers


numbers = [1, 2, 3, 4, 5, 6]

print("Even numbers:", get_even_numbers(numbers))
```

```python
# Assessment 8 - == vs is

# == checks if two values are equal.
# is checks if two variables refer to the same object.

a = [1, 2, 3]
b = [1, 2, 3]

print("a == b:", a == b)
print("a is b:", a is b)
```

## Assessment 12

```python
# Assessment 12 - for vs while loops

# A for loop goes through a range or collection.

print("For loop:")

for number in range(1, 6):
    print(number)


# A while loop continues while a condition is True.

print("While loop:")

number = 1

while number <= 5:
    print(number)
    number = number + 1
```

## Assessment 14

```python
# Assessment 14 - Palindrome

# A palindrome reads the same forwards and backwards.

word = input("Enter a word: ")

reverse = word[::-1]

if word == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")
```
